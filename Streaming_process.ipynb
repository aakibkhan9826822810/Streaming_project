```python
order_process = spark.readStream.format("delta").table('workspace.new_data.orders')
```

```python
from pyspark.sql.functions import col
orders = (
    order_process.withColumn('amount',col('price')*col('quantity'))
)
```

```python
df = (
    orders
    .writeStream
    .format("delta")
    .outputMode('append')
    .option("checkpointLocation", "/Volumes/workspace/new_data/checkpoints/orders_stream")
    .trigger(availableNow=True).
    table("workspace.new_data.orders_stream")
)
```
