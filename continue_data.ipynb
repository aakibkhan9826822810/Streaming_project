```python
df = spark.table("workspace.new_data.customers").select('customer_id')
df2 = spark.table("workspace.new_data.products").select('product_id','price')
```

```python
import pyspark.sql.functions as f


products=df2.withColumn('id',f.floor(f.rand()*100)).limit(50)
customers=df.withColumn('id',f.floor(f.rand()*100)).limit(50)

order = ( customers.join(products,on='id',how='inner').drop('id')
        .withColumn('Quantity',f.floor(f.rand()*10))
        .withColumn('order_date',f.current_timestamp())
)


order.write.mode('append').saveAsTable('workspace.new_data.orders')
```
